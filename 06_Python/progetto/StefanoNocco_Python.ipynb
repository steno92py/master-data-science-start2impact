## 2.2 Definizione della Classe Element

La classe **Element** rappresenta in modo strutturato un singolo “elemento” del sistema di gestione, pensato per raccogliere informazioni su una località (o qualunque altra entità simile) dotata di un nome, un paese di appartenenza e una lista di attrazioni. Potenzialmente può essere ingrandita a piacere per contenere altri valori. 
Lo scopo principale di questa Classe è quello di incapsulare i dati relativi a un’unità informativa (in questo caso ho inserito nome, paese, attrazioni ma è ampliabile).
Garantisce che i dati siano sempre consistenti, grazie a controlli di tipo in fase di creazione e modifica.

### Attributi chiave

- _name (str): nome dell’elemento.

- _country (str): paese di appartenenza.

- _attractions (List[str]): elenco delle attrazioni associate.

### Meccanismi di accesso e modifica

- Proprietà (@property) per leggere name, country e ottenere una copia controllata di attractions.

- Setter che validano il tipo di dato prima di aggiornare name e country.

### Metodi e attributi principali implementati

- __init__(self, name: str, country: str, attractions: List[str])
Costruttore che inizializza un nuovo elemento. Solleva TypeError se name o country non sono stringhe, o se attractions non è una lista di stringhe.
Memorizza i valori in attributi privati (_name, _country, _attractions) per proteggere l’integrità dei dati.

- @property name(self) -> str / @name.setter. Getter e setter per il nome dell’elemento. Il setter convalida che il nuovo valore sia una stringa, altrimenti solleva TypeError.

- @property country(self) -> str / @country.setter. Analoghi getter e setter per il paese, con la stessa logica di validazione del nome.

- @property attractions(self) -> List[str]. Restituisce una copia della lista delle attrazioni. Questo impedisce modifiche accidentali della lista interna, mantenendo l’incapsulamento.

- add_attraction(self, attraction: str) -> None, Aggiunge una nuova attrazione alla lista:
   1. Verifica che attraction sia una stringa, altrimenti solleva TypeError.
   2. Appende il nuovo elemento solo se valido.

- remove_attraction(self, attraction: str) -> None. Rimuove un’attrazione già presente:
   1. Se il nome non è nella lista, solleva ValueError per notificare che l’operazione non è andata a buon fine.

- __repr__(self) -> str. Rappresentazione ufficiale per il debugging: Utile per log e diagnosi.

- __eq__(self, other: object) -> bool. Definisce l’uguaglianza tra due Element:
   1. due istanze sono considerate uguali se stesso nome, stesso paese e stesse attrazioni (stessa lista e stesso ordine).
   2. Ritorna NotImplemented se other non è un Element.

```python
class Element:
    """
    Rappresenta un elemento con nome, paese e attrazioni.
    """

    def __init__(self, name: str, country: str, attractions: List[str]) -> None:
        """
        Inizializza un nuovo elemento.
        
        :param name: nome dell'elemento
        :param country: paese di riferimento
        :param attractions: lista di attrazioni (stringhe)
        :raise TypeError: se i tipi non sono corretti
        """
        # Validazione tipi
        if not isinstance(name, str):
            raise TypeError("Il nome deve essere una stringa")
        if not isinstance(country, str):
            raise TypeError("Il paese deve essere una stringa")
        if not (isinstance(attractions, list) and all(isinstance(a, str) for a in attractions)):
            raise TypeError("Le attrazioni devono essere una lista di stringhe")
        
        # Attributi privati
        self._name = name
        self._country = country
        self._attractions = attractions.copy()

    @property
    def name(self) -> str:
        """Nome dell'elemento."""
        return self._name

    @name.setter
    def name(self, value: str) -> None:
        """Modifica il nome, con validazione."""
        if not isinstance(value, str):
            raise TypeError("Il nome deve essere una stringa")
        self._name = value

    @property
    def country(self) -> str:
        """Paese di riferimento."""
        return self._country

    @country.setter
    def country(self, value: str) -> None:
        """Modifica il paese, con validazione."""
        if not isinstance(value, str):
            raise TypeError("Il paese deve essere una stringa")
        self._country = value

    @property
    def attractions(self) -> List[str]:
        """Restituisce una copia della lista di attrazioni."""
        return self._attractions.copy()

    def add_attraction(self, attraction: str) -> None:
        """
        Aggiunge una nuova attrazione.
        
        :param attraction: nome dell'attrazione
        :raises TypeError: se non è una stringa
        """
        if not isinstance(attraction, str):
            raise TypeError("L'attrazione deve essere una stringa")
        self._attractions.append(attraction)

    def remove_attraction(self, attraction: str) -> None:
        """
        Rimuove un'attrazione esistente.
        
        :param attraction: nome dell'attrazione da rimuovere
        :raises ValueError: se l'attrazione non è presente
        """
        try:
            self._attractions.remove(attraction)
        except ValueError:
            raise ValueError(f"Attrazione '{attraction}' non presente")

    def __repr__(self) -> str:
        """
        Rappresentazione ufficiale dell'oggetto (per debugging).
        """
        return (f"Element(cityName={self._name!r}, country={self._country!r}, "
                f"attractions={self._attractions!r})")

    def __eq__(self, other: object) -> bool:
        """
        Confronto tra due elementi (uguali se name, country e attrazioni coincidono).
        """
        if not isinstance(other, Element):
            return NotImplemented
        return (
            self._name == other._name and
            self._country == other._country and
            self._attractions == other._attractions
        )
```

## 2.3 Definizione della Classe SGE

La classe **SGE** (Sistema di Gestione degli Elementi) è il “motore” che organizza, conserva e fornisce operazioni su una collezione di istanze di Element. Insieme ai comandi della CLI (che ho definito in seguito), permette di gestire facilmente l’intero ciclo di vita dei record (CRUD, ovvero Create, Read, Update, Delete), fare ricerche, applicare filtri avanzati e ottenere statistiche aggregate.

### Panoramica dei metodi
- __init__(self) -> None
Inizializza una nuova collezione vuota. Internamente mantiene una lista privata _elements: List[Element].

- add_element(self, element: Element) -> None
Aggiunge un singolo Element alla collezione. Solleva TypeError se l’argomento non è un’istanza di Element.

- remove_element(self, element: Element) -> None Rimuove l’istanza indicata. Solleva ValueError se l’elemento non è presente.

- update_element(self, index: int, **attrs) -> None
Modifica uno o più attributi (name, country o attractions) dell’elemento in posizione index. Solleva IndexError se l’indice è fuori intervallo, TypeError se passi chiavi non valide.

- view_elements(self) -> None
Stampa a schermo tutti gli elementi in formato tabellare, con colonne allineate per indice, nome, paese e attrazioni.

- search_elements(self, **criteria) -> List[Element]
Restituisce la lista di elementi il cui attributo coincide esattamente con i criteria forniti (es. country="Italia"). Solleva TypeError se cerchi un campo inesistente.

- filter_elements(self, filters: List[Callable[[Element], bool]]) -> List[Element]
Ricerca avanzata applicando più funzioni booleane a ciascun Element. Restituisce solo quelli che passano tutti i filtri.

- __len__(self) -> int
Permette di usare len(sge) per ottenere il numero totale di elementi.

- __iter__(self) -> Iterator[Element]
Supporta l’iterazione diretta: for el in sge.

- clear(self) -> None
Svuota completamente la collezione.

- get_statistics(self) -> dict
Restituisce un dizionario con statistiche di base:
   1. Record totali
   2. Distribuzione città per paese
   3. Città con più attrazioni (per paese)
   4. Numero totale di attrazioni (per paese)

```python
from typing import List, Callable, Iterator

class SGE:
    """
    Sistema di Gestione degli Elementi (SGE): opera su istanze di Element,
    offrendo operazioni CRUD, ricerca semplice, filtraggio avanzato, statistiche
    e supporto a len() e iterazione.
    """

    def __init__(self) -> None:
        """
        Inizializza la collezione interna vuota.
        """
        self._elements: List[Element] = []

    def add_element(self, element: Element) -> None:
        """
        Aggiunge un nuovo elemento alla collezione.
        
        :param element: istanza di Element
        :raises TypeError: se non è un Element
        """
        if not isinstance(element, Element):
            raise TypeError("add_element richiede un'istanza di Element")
        self._elements.append(element)

    def add_elements(self, elements):
        """
        Aggiunge una lista di elementi in batch.
        :param elements: lista di istanze Element
        :raises TypeError: se la lista o i suoi elementi non sono validi
        """
        if not isinstance(elements, list):
            raise TypeError("add_elements richiede una lista di Element")
        for el in elements:
            self.add_element(el)

    def remove_element(self, element: Element) -> None:
        """
        Rimuove un elemento esistente dalla collezione.
        
        :param element: istanza di Element da rimuovere
        :raises ValueError: se l'elemento non è presente
        """
        try:
            self._elements.remove(element)
        except ValueError:
            raise ValueError("Elemento non presente nella collezione")

    def update_element(self, index: int, **attrs) -> None:
        """
        Modifica attributi di un elemento esistente, identificato dal suo indice.
        
        :param index: indice dell'elemento nella lista
        :param attrs: coppie chiave=valore per attributi di Element (name, country, attractions)
        :raises IndexError: indice non valido
        :raises TypeError: attributo inesistente o indice non intero
        """
        if not isinstance(index, int):
            raise TypeError("L'indice deve essere un intero")
        if index < 0 or index >= len(self._elements):
            raise IndexError("Indice fuori intervallo")
        element = self._elements[index]
        for attr, val in attrs.items():
            if not hasattr(element, attr):
                raise TypeError(f"{attr!r} non è un attributo di Element")
            setattr(element, attr, val)

    def view_elements(self) -> None:
        """
        Stampa tutti gli elementi in tabella:
          - Indice
          - Name
          - Country
          - Attractions (separati da virgola)
        """
        elems = self._elements
        if not elems:
            print("Collezione vuota.")
            return

        # Prepazione delle colonne
        names    = [e.name    for e in elems]
        countries= [e.country for e in elems]
        atts     = [", ".join(e.attractions) for e in elems]

        # Larghezze dinamiche
        idx_w     = len(str(len(elems)-1))
        name_w    = max(len("Name"),    *(len(n) for n in names))
        country_w = max(len("Country"), *(len(c) for c in countries))
        att_w     = max(len("Attractions"), *(len(a) for a in atts))

        # Header
        header = f"{'Id':>{idx_w}}  {'Name':<{name_w}}  {'Country':<{country_w}}  {'Attractions':<{att_w}}"
        sep    = "-" * len(header)
        print(header)
        print(sep)

        # Righe
        for idx, (n, c, a) in enumerate(zip(names, countries, atts)):
            row = f"{idx:>{idx_w}}  {n:<{name_w}}  {c:<{country_w}}  {a:<{att_w}}"
            print(row)

    def search_elements(self, **criteria) -> List[Element]:
        """
        Ricerca semplice per attributi con confronto esatto.
        
        :param criteria: es. country="Italia", name="Roma"
        :returns: lista di Element che soddisfano tutti i criteri
        :raises TypeError: attributo critero non valido
        """
        results = self._elements
        for attr, val in criteria.items():
            if not hasattr(Element, attr):
                raise TypeError(f"{attr!r} non è un attributo di Element")
            results = [el for el in results if getattr(el, attr) == val]
        return results

    def filter_elements(self, filters: List[Callable[[Element], bool]]) -> List[Element]:
        """
        Filtraggio avanzato: applica una lista di funzioni boolean al singolo elemento.
        
        :param filters: lista di callable che ricevono un Element e ritornano bool
        :returns: lista di Element per cui tutte le funzioni restituiscono True
        :raises TypeError: se uno dei filtri non è callable
        """
        if not all(callable(f) for f in filters):
            raise TypeError("Tutti i filtri devono essere funzioni che ritornano bool")
        return [el for el in self._elements if all(f(el) for f in filters)]

    def __len__(self) -> int:
        """
        Permette di usare len(sge) per ottenere il numero di elementi.
        """
        return len(self._elements)

    def __iter__(self) -> Iterator[Element]:
        """
        Supporta l'iterazione for el in sge.
        """
        return iter(self._elements)

    def clear(self) -> None:
        """
        Svuota completamente la collezione.
        """
        self._elements.clear()

    def get_statistics(self) -> dict:
        """
        Calcola e stampa statistiche di base sugli elementi presenti nella collezione.

        Le sezioni stampate sono:
          1. Record totali: numero totale di elementi nella collezione.
             - Formato: "Record totali: {total}\n"

          2. Distribuzione città per paese:
             - Per ogni paese viene stampato:
               "  {paese}: {conteggio_elementi}"
             - Preceduto dalla riga di intestazione:
               "Distribuzione città per paese:"

          3. Città con più attrazioni per paese:
             - Per ogni paese viene stampato:
               "  {paese}: {nome_città} ({numero_attrazioni} attrazioni)"
             - Preceduto dalla riga di intestazione:
               "\nCittà con più attrazioni per paese:"

          4. Numero totale di attrazioni per paese:
             - Per ogni paese viene stampato:
               "  {paese}: {numero_totale_attrazioni}"
             - Preceduto dalla riga di intestazione:
               "\nNumero totale di attrazioni per paese:"

        Se nella collezione non ci sono elementi, viene stampato:
          "Nessun elemento nella collezione."
        e il metodo termina senza stampare ulteriori sezioni.

        :noargs:
        :raises None: non solleva eccezioni; se la collezione è vuota, si limita a stampare il messaggio di avviso.
        :example:
        """
        elems = self._elements
        if not elems:
            print("Nessun elemento nella collezione.")
            return

        # Calcoli
        total = len(elems)
        distrib_citta = {}
        total_atts = {}
        max_atts_city = {}

        for el in elems:
            c = el.country
            # Conteggio città per paese
            distrib_citta[c] = distrib_citta.get(c, 0) + 1
            # Totale attrazioni per paese
            total_atts[c]   = total_atts.get(c, 0) + len(el.attractions)
            # Individua città con max attrazioni
            current_max = max_atts_city.get(c)
            if (current_max is None) or (len(el.attractions) > len(current_max[1])):
                max_atts_city[c] = (el.name, el.attractions)

        # Stampa una volta sola per ciascuna sezione
        print(f"Record totali: {total}\n")

        print("Distribuzione città per paese:")
        for paese, cnt in distrib_citta.items():
            print(f"  {paese}: {cnt}")

        print("\nCittà con più attrazioni per paese:")
        for paese, (city, atts) in max_atts_city.items():
            print(f"  {paese}: {city} ({len(atts)} attrazioni)")

        print("\nNumero totale di attrazioni per paese:")
        for paese, n in total_atts.items():
            print(f"  {paese}: {n}")


    def save_to_file(self, filepath):
        """
        Salva tutti gli elementi correnti su un file in formato JSON.

        Ogni elemento viene serializzato come un dizionario contenente:
          - "name": nome dell'Element (string)
          - "country": paese dell'Element (string)
          - "attractions": lista di attrazioni (List[str])

        Il JSON risultante è una lista di dizionari, indentata di 2 spazi e
        senza caratteri ASCII di escape (ensure_ascii=False).

        :param filepath: percorso completo del file (incluso nome) dove salvare i dati.
                         Deve essere una stringa valida.
        :raises TypeError: se `filepath` non è una stringa.
        :raises IOError: se ci sono problemi di scrittura (permessi, disco pieno, ecc.)
        :example:

            sge = SGE()
            # … aggiungo qualche elemento …
            sge.save_to_file("dati_sge.json")
            # Output atteso: Dati salvati su 'dati_sge.json'.
        """
        data = []
        for el in self._elements:
            data.append({
                "name": el.name,
                "country": el.country,
                "attractions": el.attractions
            })
        with open(filepath, 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
        print(f"Dati salvati su '{filepath}'.")


    def load_from_file(self, filepath):
        """
        Carica gli elementi da un file JSON e sostituisce la collezione interna.

        Il file JSON deve contenere una lista di oggetti (dizionari), ciascuno con le chiavi:
          - "name": stringa non vuota
          - "country": stringa non vuota
          - "attractions": lista di stringhe (anche vuota)

        Ogni riga (oggetto) viene validata in termini di presenza di chiavi e tipi:
          - Se il formato non è una lista di oggetti, solleva ValueError.
          - Se un singolo oggetto manca di una delle chiavi, solleva ValueError specificando l'indice.
          - Se i tipi delle singole proprietà non sono corretti, solleva TypeError specificando l'indice.

        Prima di aggiungere i nuovi elementi, la collezione interna viene svuotata,
        chiamando `self.clear()`.

        :param filepath: percorso completo del file JSON da leggere.
                         Deve essere una stringa valida.
        :raises TypeError: se `filepath` non è una stringa, o se i tipi dentro il JSON non corrispondono.
        :raises ValueError: se il file non contiene una lista di oggetti o se un oggetto è malformato.
        :raises IOError: se ci sono problemi di lettura (file inesistente, permessi, formato JSON invalido, ecc.)
        :example:

            sge = SGE()
            sge.load_from_file("dati_sge.json")
            # Output atteso: Dati caricati da 'dati_sge.json' e collezione aggiornata.
        """
        with open(filepath, 'r', encoding='utf-8') as f:
            data = json.load(f)
        if not isinstance(data, list):
            raise ValueError("Formato del file non valido: attesa lista di oggetti.")
        self.clear()
        for idx, item in enumerate(data):
            if not all(k in item for k in ("name", "country", "attractions")):
                raise ValueError(f"Dato mancante in riga {idx}: {item}")
            name = item["name"]
            country = item["country"]
            atts = item["attractions"]
            if not (isinstance(name, str) and isinstance(country, str) and isinstance(atts, list)):
                raise TypeError(f"Tipi non validi in riga {idx}: {item}")
            el = Element(name, country, atts)
            self.add_element(el)
        print(f"Dati caricati da '{filepath}' e collezione aggiornata.")
```

# 2.3 Definizione della Classe SGECLI (Command Line Interface)

La classe **SGECLI** (SGE Command-Line Interface) fornisce un’interfaccia testuale interattiva per gestire da terminale la collezione di Element tramite il motore SGE. Permette all’utente di eseguire tutte le operazioni principali (aggiunta, visualizzazione, ricerca, filtraggio, statistiche) senza scrivere codice Python, semplicemente digitando comandi. Non era richiesta dal progetto ma mi sono divertito assai nella realizzazione.

**Metodi principali e comandi**
- __init__(self, sge: SGE). Riceve un’istanza di SGE e prepara:
  1. il dizionario self.commands che mappa stringhe di comando ai metodi corrispondenti
  2. lo stato self.running per il ciclo REPL


- start(self) -> None. Avvia il loop interattivo:
  1. Mostra il prompt SGE>
  2. Legge la riga, la suddivide in comando e argomenti
  3. Esegue il metodo associato, gestendo eccezioni e il comando di uscita


- prompt_or_back(self, text: str) -> str. Wrapper di input() che:
  1. legge l’utente
  2. Solleva ReturnToMain se viene digitato back (case-insensitive)
  3. Utile per interrompere qualunque dialogo multi-step e tornare al menu principale.


- do_help(self, args: List[str]) -> None. Stampa la guida ai comandi, con descrizioni dettagliate ed esempi d’uso.

- do_add(self, args: List[str]) -> None. Aggiunge o aggiorna un elemento:
  1. chiede name, country, attrazioni (con prompt_or_back)
  2. Se trova già un record con stesso name+country, unisce le attrazioni evitando duplicati
  3. Altrimenti, crea un nuovo Element

- do_add_many(self, args: List[str]) -> None. Permette di inserire in sequenza più elementi:
  1. chiede quanti
  2. richiama do_add per ciascuno

- do_view(self, args: List[str]) -> None. Richiama sge.view_elements(), mostrando la tabella allineata di tutti gli elementi.

- do_search(self, args: List[str]) -> None. Ricerca semplice per attributo:
  1. Sintassi: search <campo> <valore>
  2. Stampa tutti gli Element che hanno l’attributo (name o country) esattamente uguale al valore.

- do_filter(self, args: List[str]) -> None. Filtraggio avanzato interattivo:
  1. chiede quanti filtri applicare
  2. per ciascuno, lettura di campo (name/country/attractions), operatore (equals/contains), valore
  3. costruisce una lista di funzioni (Element → bool)
  4. esegue sge.filter_elements(...) e mostra i risultati

- do_stats(self, args: List[str]) -> None. Richiama sge.view_statistics(), stampando:
  1. Record totali
  2. Distribuzione città per paese
  3. Città con più attrazioni per paese
  4. Numero totale di attrazioni per paese

- do_save(self, args: text: str). Salva i dati in un file .json

- do_load(self, args: text: str). Carica i dati in un file .json

- do_exit(self, args: List[str]) -> None. Imposta self.running = False e termina il loop, chiudendo la CLI.

```python
# Interfaccia a Riga di Comando (CLI) per SGE

class SGECLI:
    """
    CLI interattiva per gestire la collezione di elementi con SGE.
    
    Comandi:
      - help                    : mostra questa guida
      - add                     : aggiunge un nuovo elemento
      - add_many                : aggiunge elementi multipli
      - view                    : visualizza tutti gli elementi
      - search <campo> <valore> : ricerca per attributo (es. search country Italia)
      - filter                  : placeholder per filtraggio avanzato
      - stats                   : mostra statistiche sugli elementi
      - save <filepath>         : Salva nella posizione selezionata
      - load <filepath>         : Carica il file dalla posizione selezionata
      - exit                    : esci dall'interfaccia
    """
    
     # Logo dell'azienda ORIZON (alt+219 per il full block)
    LOGO = """ 
████████████████████████████████████████████    
█  ███   ████   █████  █████   ███   █   █ █
█ █   █  █   █    █       █   █   █  ██  █ █
█ █   █  ████     █      █    █   █  █ █ █ █
█ █   █  █  █     █     █     █   █  █  ██ █
█  ███   █   █  █████  █████   ███   █   █ █
████████████████████████████████████████████
╔══════════════════════════════════════════╗
║           MAIN MENU SGE ORIZON           ║
║           Developed by steno.py          ║
╚══════════════════════════════════════════╝
"""
    MY_LOGO="""
████████████████████████████████████████████
█ █████ █████ ████ █   █  ███    ████  █ █ █
█ █       █   █    ██  █ █   █   █   █ █ █ █
█ █████   █   ████ █ █ █ █   █   ████  ███ █
█     █   █   █    █  ██ █   █   █      █  █
█ █████   █   ████ █   █  ███  █ █      █  █
████████████████████████████████████████████  
    """

    def __init__(self, sge):
        """
        :param sge: istanza di SGE già inizializzata
        """
        self.sge = sge
        self.running = False
        self.commands = {
            'help': self.do_help,
            'add': self.do_add,
            'add_many': self.do_add_many,
            'view': self.do_view,
            'search': self.do_search,
            'filter': self.do_filter,
            'stats': self.do_stats,
            'exit': self.do_exit,
            'save': self.do_save,
            'load': self.do_load
        }


    class ReturnToMain(Exception):
        """Eccezione usata per uscire da un comando e tornare al menu."""
        pass

    def prompt_or_back(self, text: str) -> str:
        """
        Legge un valore da console.
        Se l’utente digita 'back' (case-insensitive), solleva ReturnToMain.
        """
        val = input(text).strip()
        if val.lower() == 'back':
            raise self.ReturnToMain()
        return val

    def start(self):
        """Avvia il loop interattivo."""
        print(self.LOGO)
        print("Benvenuto nel Sistema di Gestione degli Elementi di Orizon. \nDigita 'help' per i comandi disponibili.\n")
        self.running = True
        while self.running:
            try:
                line = input("SGE> ").lower().strip()
            except (EOFError, KeyboardInterrupt):
                print("\nArrivederci!")
                break
            if not line:
                continue
            parts = line.split()
            cmd, *args = parts
            action = self.commands.get(cmd)
            if not action:
                print(f"Comando sconosciuto: {cmd}")
                continue
    
            try:
                action(args)
            except self.ReturnToMain:
                # l’utente ha digitato 'back'
                continue
            except Exception as e:
                print(f"Errore: {e}")


    def do_help(self, args):
        """Mostra la guida ai comandi."""
        print(""" 
╔══════════════════════════════════════════╗
║             GUIDA AI COMANDI             ║
╚══════════════════════════════════════════╝
""")
        print(self.__doc__)


    def do_add(self, args):
        """Aggiunge un nuovo elemento interattivamente."""
        print(""" 
╔══════════════════════════════════════════╗
║          CARICA SINGOLO RECORD           ║
╚══════════════════════════════════════════╝
""")
        print("""      
        Aggiunge un nuovo elemento interattivamente.
        Vengono richiesti:
        - Nome Città: stringa
        - Paese     : stringa
        - Attrazioni: elenco di nomi separati da virgola\n
        Funzionamento:
        - Se esiste un elemento con stesso name e country, unisce le attrazioni
        evitando duplicati.
        - Altrimenti, crea un nuovo elemento.
        - Supporta 'back' per uscire.\n""")
        
        try:
            name = self.prompt_or_back("Nome Città (o 'back' per uscire): ")
            country = self.prompt_or_back("Paese (o 'back'): ")
            raw = self.prompt_or_back("Attrazioni (separate da virgola, o 'back'): ")
        except self.ReturnToMain:
            print("Operazione interrotta, ritorno al menu.")
            print(self.LOGO)
            print("Benvenuto nel Sistema di Gestione degli Elementi di Orizon. \nDigita 'help' per i comandi disponibili.\n")
            return
            
            # Preparo lista attrazioni senza duplicati interni
        new_atts = [a.strip() for a in raw.split(',') if a.strip()]
        
        # Cerco elemento esistente
        existing = self.sge.search_elements(name=name, country=country)
        if existing:
            el = existing[0]
            added = 0
            for a in new_atts:
                if a not in el.attractions:
                    el.add_attraction(a)
                    added += 1
            if added:
                print(f"Aggiornato '{name}' ({country}): aggiunte {added} nuove attrazioni.")
            else:
                print(f"Nessuna nuova attrazione da aggiungere per '{name}' ({country}).")
        else:
            # Nessun elemento uguale, ne creo uno nuovo
            el = Element(name, country, new_atts)
            self.sge.add_element(el)
            print(f"Elemento '{name}' ({country}) aggiunto con successo.")

    def do_add_many(self, args):
        """
        Comando add_many: chiede quanti elementi inserire e richiama do_add in loop.
        """
        print(""" 
╔══════════════════════════════════════════╗
║          CARICA PIU' RECORD              ║
╚══════════════════════════════════════════╝
""")
        try:
            n = self.prompt_or_back("Quanti elementi vuoi aggiungere? (o 'back' per uscire): ")
            n = int(n)
        except self.ReturnToMain:
            print("Operazione interrotta, ritorno al menu.")
            print(self.LOGO)
            print("Benvenuto nel Sistema di Gestione degli Elementi di Orizon. \nDigita 'help' per i comandi disponibili.\n")
            return
        except ValueError:
            print("Numero non valido.")
            return
        for i in range(n):
            print(f""" 
╔══════════════════════════════════════════╗
     -- Inserimento elemento {i+1}/{n} --   
╚══════════════════════════════════════════╝
""")
            self.do_add([])
        print(f"{n} elementi aggiunti.")


    def do_view(self, args):
        """Visualizza tutti gli elementi."""
        print(""" 
╔══════════════════════════════════════════╗
║           LISTA DEI RECORD               ║
╚══════════════════════════════════════════╝
""")
        self.sge.view_elements()


    def do_search(self, args):
        """
        Ricerca per attributo.
        Uso: search <campo> <valore>
        """
        print(""" 
╔══════════════════════════════════════════╗
║         RICERCA <CAMPO> <VALORE>         ║
╚══════════════════════════════════════════╝
""")
        if len(args) < 2:
            print("Uso corretto: search <campo> <valore>")
            return
        field = args[0]
        value = " ".join(args[1:])
        try:
            results = self.sge.search_elements(**{field: value})
        except TypeError as e:
            print(f"Errore di ricerca: {e}")
            return
        if results:
            for el in results:
                print(el)
        else:
            print("Nessun elemento trovato.")


    def do_filter(self, args):
        """
        Filtraggio avanzato interattivo:
        consente di specificare più condizioni per cityName, Country o Attractions.
        Digita 'back' in qualsiasi prompt per tornare al menu principale.
        """
        print(""" 
╔══════════════════════════════════════════╗
║             FILTRO AVANZATO              ║
╚══════════════════════════════════════════╝
""")
        try:
            # Numero di filtri da applicare
            n_raw = self.prompt_or_back("Quanti filtri vuoi applicare? ")
            n = int(n_raw)
        except self.ReturnToMain:
            print("Operazione annullata, ritorno al menu.")
            print(self.LOGO)
            return
        except ValueError:
            print("Numero non valido. Uso: un intero maggiore di 0.")
            return

        filters = []
        for i in range(1, n+1):
            try:
                campo = self.prompt_or_back(f"Filtro #{i} - campo (name/country/attractions): ").lower()
                if campo not in ("name", "country", "attractions"):
                    print(f"Campo '{campo}' non valido.")
                    return

                # Definiamo possibili operatori
                ops = "equals/contains" if campo in ("name", "country") else "contains"
                operatore = self.prompt_or_back(f"Operatore ({ops}): ").lower()
                valore = self.prompt_or_back("Valore di confronto: ")

            except self.ReturnToMain:
                print("Operazione annullata, ritorno al menu.")
                return

            # Costruzione della funzione di filtro
            if campo in ("name", "country"):
                if operatore == "equals":
                    filters.append(lambda el, c=campo, v=valore: getattr(el, c) == v)
                elif operatore == "contains":
                    filters.append(lambda el, c=campo, v=valore: v in getattr(el, c))
                else:
                    print(f"Operatore '{operatore}' non valido per campo '{campo}'.")
                    return
            else:  # attractions
                if operatore == "contains":
                    filters.append(lambda el, v=valore: v in el.attractions)
                else:
                    print(f"Operatore '{operatore}' non valido per 'attractions'.")
                    return

        # Applicazione dei filtri e stampa risultati
        results = self.sge.filter_elements(filters)
        if results:
            print(f"Risultati ({len(results)}):")
            for el in results:
                print(f"  - {el}")
        else:
            print("Nessun elemento soddisfa i criteri.")

    
    def do_stats(self, args):
        """Mostra le statistiche della collezione."""
        print(""" 
╔══════════════════════════════════════════╗
║               STATISTICHE                ║
╚══════════════════════════════════════════╝
""")
        stats = self.sge.get_statistics()
    


    
    def do_save(self, args):
        """
        Salva l’intera collezione su un file JSON.
        Uso: save <filepath>
        """
        print(""" 
╔══════════════════════════════════════════╗
║           SALVA IL DATASET               ║
╚══════════════════════════════════════════╝
""")
        if len(args) != 1:
            print("Uso corretto: save <filepath>")
            return
        filepath = args[0]
        try:
            # self.sge è l’istanza di SGE passata a SGECLI.__init__
            self.sge.save_to_file(filepath)
            print("Dataset salvato con successo!")
        except Exception as e:
            print(f"Errore durante il salvataggio: {e}")

    def do_load(self, args):
        """
        Carica la collezione da un file JSON precedentemente salvato.
        Uso: load <filepath>
        """
        print(""" 
╔══════════════════════════════════════════╗
║           CARICA IL DATASET              ║
╚══════════════════════════════════════════╝
""")
        if len(args) != 1:
            print("Uso corretto: load <filepath>")
            return
        filepath = args[0]
        try:
            self.sge.load_from_file(filepath)
            print("Dataset caricato con successo!")
        except Exception as e:
            print(f"Errore durante il caricamento: {e}")

    def do_exit(self, args):
        """Esce dall'interfaccia."""
         # Chiedi se salvare i dati
        risposta = input("Vuoi salvare i dati prima di uscire? (Si/No): ").strip().lower()
        if risposta == "si":
            # Se l'utente vuole salvare, chiedi il percorso del file
            filepath = input("Inserisci il percorso del file per salvare i dati (es. dati.json): ").strip()
            try:
                self.sge.save_to_file(filepath)
                print("Dataset salvato con successo!")
            except Exception as e:
                print(f"Errore durante il salvataggio: {e}")
        print("Arrivederci!")
        print(self.MY_LOGO)
        self.running = False
```
